# 08_불_비교_논리연산

오른쪽 위 커널은 **Python 학습 (.venv)**를 선택해. 위에서 아래로 실행해. `input()` 셀은 화면 맨 위 입력창에서 값을 기다리는 것이 정상이고, 중단하려면 ■ 버튼을 눌러.


```python
# 불과 비교 연산자 사용하기
print(True)
print(type(True))
print(False)
print(type(False))
```

```python
# 비교연산의 판단결과
print(3 > 1)
print(1 > 3)
```

```python
# 숫자가 같은지 다른지 비교
print(10 == 10)
print(10 != 10)
```

```python
print(10 > 20)
print(10 <= 10)
```

```python
# 논리연산자 사용하기
print(True and True)
print(True and False)
print(False and True)
print(False and False)
```

```python
# or 논리연산자 사용하기
print(True or True)
print(True or False)
print(False or True)
print(False or False)
```

```python
# not

print(not True and False or not False)
```

```python
# 논리 연산자와 비교연산자를 함께 사용하기
print(10 == 10 and 10 != 5)
print( (10 > 5) or (10 < 3) )
```

```python
# ==는 값 비교, is는 같은 객체인지 비교해.
a = [1, 2]
b = [1, 2]
print(a == b)  # True: 값이 같아.
print(a is b)  # False: 서로 다른 리스트야.
```

```python
korean = 92
english = 47
math = 86
science = 81
print(korean >= 50 and english >= 50 and math >= 50 and science >= 50)
```
