```python
# 딕셔너리 만들기

lux = {"체력": 490, "마나": 334}

print(lux["체력"])
```

```python
lux = dict(health=490, mana=334, melee=550, armor=18.72)

print(lux)
print(lux["mana"])
```

```python
lux2 = dict(zip(["health", "mana", "melee", "armor"], [490, 334, 550, 18.72]))

lux2
```

```python
# 딕셔너리의 키에 접근하고 값 할당하기

lux = dict(zip(["health", "mana", "melee", "armor"], [490, 334, 550, 18.72]))

print(lux["health"])  # 490
print(lux["armor"])  # 18.72
```

```python
lux = dict(zip(["health", "mana", "melee", "armor"], [490, 334, 550, 18.72]))

lux["health"] = 2037  # 체력 변경
lux["mana"] = 1184  # 마나 변경

print(lux)
```
