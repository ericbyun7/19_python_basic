```python
# if: 조건이 맞으면
# else: 그렇지 않으면

x = 5

if x == 10:
    print("10입니다.")
else:
    print("10이 아닙니다.")

# 결과: 10이 아닙니다.
# 생각해 보기: x를 10으로 바꾸면?
# 정답: 10입니다.
```

```python
# if와 else는 같은 위치에 맞춰.
# 각각의 실행 코드는 공백 4칸을 넣어.
# else 뒤에는 조건 없이 :만 붙여.

x = 10

if x == 10:
    print("조건이 맞아요.")
    print("if 안입니다.")
else:
    print("조건이 틀려요.")
    print("else 안입니다.")

print("이 줄은 항상 실행돼요.")
```

```python
# False, None, 숫자 0, 빈 문자열은 거짓으로 판단해.
# 0이 아닌 숫자와 내용이 있는 문자열은 참이야.

x = 0

if x:
    print("참")
else:
    print("거짓")

# 결과: 거짓
# 생각해 보기: x를 -5로 바꾸면?
# 정답: 참. 음수도 0이 아니니까!
```

```python
# and: 조건이 모두 맞아야 참
# or: 조건 중 하나라도 맞으면 참
# not: 참과 거짓을 뒤집음

x = 15

if x > 0 and x < 20:
    print("0보다 크고 20보다 작아요.")
else:
    print("범위를 벗어났어요.")

# 결과: 0보다 크고 20보다 작아요.

# 같은 조건을 짧게 쓰는 방법:
if 0 < x < 20:
    print("같은 뜻이에요.")
```

```python
# 14.6 연습문제: 합격 여부 판단하기

written_test = 75  # 필기 점수
coding_test = True  # 코딩 시험 통과

# and는 두 조건이 모두 맞아야 해.
if written_test >= 80 and coding_test == True:
    print("합격")
else:
    print("불합격")
```

```python
# 14.7 심사문제: 합격 여부 판단하기

# 네 과목 점수를 공백으로 구분해서 입력
korean, english, mathematics, science = map(int, input().split())

# 모든 점수가 0~100점인지 확인
if (
    0 <= korean <= 100
    and 0 <= english <= 100
    and 0 <= mathematics <= 100
    and 0 <= science <= 100
):
    average = (korean + english + mathematics + science) / 4

    if average >= 80:
        print("합격")
    else:
        print("불합격")
else:
    print("잘못된 점수")
```
