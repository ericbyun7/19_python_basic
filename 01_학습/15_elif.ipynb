```python
# if부터 위에서 아래로 확인해.

x = 20

if x == 10:
    print("10입니다.")
elif x == 20:
    print("20입니다.")

# 결과: 20입니다.
```

```python
# if: 첫 조건
# elif: 앞 조건이 틀리면 다음 조건 확인
# else: 앞 조건이 전부 틀리면 실행

x = 30

if x == 10:
    print("10입니다.")
elif x == 20:
    print("20입니다.")
else:
    print("10도 20도 아닙니다.")

# 결과: 10도 20도 아닙니다.
# else는 반드시 마지막에 써.
```

```python
# 실행하면 맨 위 입력칸에 1, 2, 3 중 하나를 입력해.

button = int(input())

if button == 1:
    print("콜라")
elif button == 2:
    print("사이다")
elif button == 3:
    print("환타")
else:
    print("제공하지 않는 메뉴")

# 생각해 보기: 2를 입력하면?
# 정답: 사이다
```

```python
x = 20

if x >= 10:
    print("10 이상")
elif x >= 20:
    print("20 이상")
else:
    print("10 미만")

# 결과: 10 이상
# 첫 조건이 맞아서 아래 elif는 확인하지 않아!
```

```python
# 15.3 연습문제

x = int(input())

if 11 <= x <= 20:
    print("11~20")
elif 21 <= x <= 30:
    print("21~30")
else:
    print("아무것도 해당하지 않음")
```

```python
# 15.4 심사문제
# 문제에서는 나이가 7세 이상으로 입력돼.

age = int(input())
balance = 9000

if age <= 12:
    balance = balance - 650
elif age <= 18:
    balance = balance - 1050
else:
    balance = balance - 1250

print(balance)
```
