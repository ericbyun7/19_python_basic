```python
# while 반복문으로 요소 출력

a = [38, 21, 53, 62, 19]

i = 0
while i < len(a):
    print(a[i])
    i += 1
```

```python
# 가장 작은 수와 가장 큰 수 구하기

a = [38, 21, 53, 62, 19]

smallest = a[0]

for i in a:
    if i < smallest:
        smallest = i
print(smallest)
```

```python
# 가장큰수 구하기

a = [38, 21, 53, 62, 19]

largest = a[0]  # 처음에는 38을 최댓값으로 정해.

for i in a:
    if i > largest:  # 더 큰 값을 만나면
        largest = i  # 최댓값을 바꿔.

print(largest)  # 62
```

```python
# 중첩 반복문으로 리스트 만들기

c = []

for i in range(2, 10):  # 2단~9단
    for j in range(1, 10):  # 각 단에서 1~9 곱하기
        c.append(i * j)

print(c)
```

```python
# 22.6 리스트에 map 사용하기
# map()으로 각 요소에 int()를 적용해 정수로 바꿔.

a = [1.2, 2.5, 3.7, 4.6]

a = list(map(int, a))

print(a)  # [1, 2, 3, 4]

# map(int, a) → 값을 하나씩 정수로 바꾸기
# list(...)  → 결과를 리스트로 만들기
# a =        → 새 리스트를 a에 저장하기

# int()는 반올림하지 않고 소수점 아래를 버려.
```

```python
# 22.7 튜플 응용하기
# 튜플은 값을 추가하거나 수정할 수 없어.

# 1. 값의 인덱스 찾기
a = (38, 21, 53, 62, 19, 53)
print(a.index(53))  # 2: 처음 나오는 위치


# 2. 값의 개수 세기
print(a.count(53))  # 2: 53이 두 개


# 3. 반복문으로 값 꺼내기
a = (38, 21, 53, 62, 19)

for i in a:
    print(i)


# 4. 튜플 표현식 대신 tuple() 사용하기
a = tuple(i for i in range(10) if i % 2 == 0)
print(a)  # (0, 2, 4, 6, 8)


# 5. map으로 실수를 정수로 바꾸기
a = (1.2, 2.5, 3.7, 4.6)
a = tuple(map(int, a))
print(a)  # (1, 2, 3, 4)


# 6. 최솟값·최댓값·합계
a = (38, 21, 53, 62, 19)

print(min(a))  # 19
print(max(a))  # 62
print(sum(a))  # 193
```

```python
# 22.7 튜플 표현식 사용하기
# 교재 예제: 0~9 중 짝수로 튜플 만들기

a = tuple(i for i in range(10) if i % 2 == 0)

print(a)  # (0, 2, 4, 6, 8)

# range(10)   → 0~9를 하나씩 꺼내기
# i % 2 == 0  → 짝수만 고르기
# tuple(...)  → 골라낸 값으로 튜플 만들기
```

```python
# 22.9 연습문제: 리스트에서 특정 요소만 뽑아내기

a = ["alpha", "bravo", "charlie", "delta", "echo", "foxtrot", "golf", "hotel", "india"]

b = [i for i in a if len(i) == 5]

print(b)
```

```python
# 22.10 심사문제

start, stop = map(int, input().split())

a = [2**i for i in range(start, stop + 1)]

del a[1]  # 두 번째 요소 삭제
del a[-2]  # 뒤에서 두 번째 요소 삭제

print(a)
```
