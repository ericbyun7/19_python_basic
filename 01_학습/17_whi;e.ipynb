```python
# 교재의 100번 반복을 먼저 3번으로 연습해 보자.

i = 0  # 시작 숫자

while i < 20:  # i가 3보다 작은 동안 반복
    print("Hello, world!", i)
    i += 1  # i를 1 증가시켜. i = i + 1

# 결과:
# Hello, world! 0
# Hello, world! 1
# Hello, world! 2
```

```python
i = 30

while i > 0:
    print(i)
    i -= 1  # i = i - 1

# 결과:
# 3
# 2
# 1
```

```python
# 실행 후 맨 위 입력칸에 3을 입력.

count = int(input())

i = 0

while i < count:
    print("안녕!")
    i += 1

# 3을 입력하면 안녕!이 3번 출력.
```
