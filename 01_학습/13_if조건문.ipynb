```python
# ── 1. if란? ──
# "만약 조건이 맞으면 실행해"

x = 10

if x == 10:
    print("10입니다.")


# ── 2. 기호 뜻 ──
# =   → 값 저장
# ==  → 같은지 비교
# :   → 조건 끝
# 공백 4칸 → 조건이 맞을 때 실행할 코드


# ── 3. 생각해 보기 ──
# 아래는 "10입니다."가 출력될까?

x = 5

if x == 10:
    print("10입니다.")

# 정답: 출력되지 않아.
# x가 5라서 조건이 틀렸기 때문이야.


# ── 4. 들여쓰기 차이 ──

x = 5

if x == 10:
    print("조건이 맞을 때만 실행")

print("이 줄은 항상 실행")
```

```python
# 13.1 조건이 맞을 때만 실행해.
# =는 저장, ==는 같은지 비교야.

x = 10

if x == 10:
    print("10입니다.")

# 결과: 10입니다.
# x를 5로 바꾸면 출력되지 않아.
```

```python
# pass는 아무 일도 하지 않고 넘어가.
# 나중에 코드를 작성할 자리에 넣어 둬.

x = 10

if x == 10:
    pass

# 결과: 출력 없음
```

```python
# 13.2 같은 if에 속한 코드는 들여쓰기를 맞춰.
# 공백 4칸을 사용하면 좋아.

x = 10

if x == 10:
    print("x에 들어있는 숫자는")
    print("10입니다.")

# 결과:
# x에 들어있는 숫자는
# 10입니다.
```

```python
# 들여쓰기가 없으면 if 바깥의 코드야.

x = 5

if x == 10:
    print("10입니다.")

print("조건과 관계없이 실행돼요.")

# 생각해 보기: 무엇이 출력될까?
# 정답: 조건과 관계없이 실행돼요.
```

```python
# 13.3 if 안에 if를 넣는 것을 중첩이라고 해.
# 안쪽으로 들어갈 때마다 공백을 4칸 더 넣어.

x = 15

if x >= 10:
    print("10 이상입니다.")

    if x == 15:
        print("15입니다.")

    if x == 20:
        print("20입니다.")

# 결과:
# 10 이상입니다.
# 15입니다.

# 바깥 조건이 틀리면 안쪽 조건도 검사하지 않아.
```

```python
# 13.4 input()으로 입력받고 int()로 정수로 바꿔.
# 실행 후 VS Code 위쪽 입력칸에 10 또는 20을 입력해.

x = int(input())

if x == 10:
    print("10입니다.")

if x == 20:
    print("20입니다.")

# 생각해 보기: 30을 입력하면?
# 정답: 두 조건 모두 틀려서 출력되지 않아.
```

```python
# 13.6 x가 10이 아닐 때 ok를 출력하기.
# 먼저 생각해 보기: "다르다"는 어떤 기호일까?
# 정답: !=

x = 5

if x != 10:
    print("ok")

# 결과: ok
```
