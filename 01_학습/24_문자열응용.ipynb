## 24.1.1 문자열 바꾸기

```python
# 24.1.1 문자열 바꾸기

s = "Hello, world!"
s = s.replace("world", "Python")
print(s)  # Hello, Python!
```

## 24.1.2 문자 바꾸기

```python
# 24.1.2 문자 바꾸기

table = str.maketrans("aeiou", "12345")
print("apple".translate(table))  # 1ppl2
```

## 24.1.3~4 나누기와 연결하기

```python
# 24.1.3~4 나누기와 연결하기

words = "apple pear grape pineapple orange".split()
print(words)
print("-".join(words))
# split은 리스트로 나누고, join은 문자열로 연결해.
```

## 24.1.5~6 대문자와 소문자

```python
# 24.1.5~6 대문자와 소문자

print("python".upper())  # PYTHON
print("PYTHON".lower())  # python
```

## 24.1.7~9 공백 삭제

```python
# 24.1.7~9 공백 삭제

s = "   Python   "
print(repr(s.lstrip()))  # 왼쪽 공백 삭제
print(repr(s.rstrip()))  # 오른쪽 공백 삭제
print(repr(s.strip()))  # 양쪽 공백 삭제
# repr은 따옴표까지 보여 줘서 공백을 확인하기 편해.
```

## 24.1.10~12 특정 문자 삭제

```python
# 24.1.10~12 특정 문자 삭제

s = ", python."
print(repr(s.lstrip(",.")))
print(repr(s.rstrip(",.")))
print(repr(s.strip(",.")))
# 지정한 문자들을 양 끝에서 지워. 중간 공백은 남아.
```

## 24.1.13~16 정렬과 메서드 연결

```python
# 24.1.13~16 정렬과 메서드 연결

print(repr("python".ljust(10)))
print(repr("python".rjust(10)))
print(repr("python".center(10)))
print(repr("python".rjust(10).upper()))
# 전체 10칸에 맞춰 정렬해. 문자열을 잘라내지는 않아.
```

## 24.1.17 앞에 0 채우기

```python
# 24.1.17 앞에 0 채우기

print("35".zfill(4))  # 0035
```

## 24.1.18~22 위치와 개수 찾기

```python
# 24.1.18~22 위치와 개수 찾기

s = "apple pineapple"
print(s.find("pl"))  # 2
print(s.rfind("pl"))  # 12
print(s.index("pl"))  # 2
print(s.rindex("pl"))  # 12
print(s.count("pl"))  # 2
print(s.find("xy"))  # -1
# index는 찾는 문자열이 없으면 ValueError가 발생해.
```

## 24.2.1~3 문자열과 숫자 넣기

```python
# 24.2.1~3 문자열과 숫자 넣기

print("I am %s." % "james")
print("I am %d years old." % 20)
print("%.2f" % 2.3)  # 2.30
# %s는 문자열, %d는 정수, %.2f는 소수점 아래 두 자리야.
```

## 24.2.4 서식 지정자로 정렬하기

`%10s`는 오른쪽 정렬, `%-10s`는 왼쪽 정렬입니다. 마이너스는 따옴표 안의 `%` 바로 뒤에 넣습니다.

```python
# 서식 지정자로 문자열 정렬

print("%10s" % "python")
print("%-10s" % "python")
```

## 24.2.5 값 여러 개 넣기

```python
# 문자열 안에 값 여러개 넣기
print("Today is %d %s" % (3, "April"))
```

## 24.2.6~10 format으로 값 넣기

```python
# 24.2.6~10 format으로 값 넣기

print("Hello, {0}".format("world!"))
print("Hello, {0} {2} {1}".format("Python", "Script", 3.6))
print("Hello, {} {} {}".format("Python", "Script", 3.6))
print("Hello, {language} {version}".format(language="Python", version=3.6))
language = "Python"
print(f"Hello, {language}")
```

## 24.2 format 정렬과 숫자 표시

```python
# 24.2 format 정렬과 숫자 표시

print(repr("{0:<10}".format("python")))
print(repr("{0:>10}".format("python")))
print(repr("{0:^10}".format("python")))
print("{0:03d}".format(1))
print("{0:.2f}".format(2.3))
print("{0:,}".format(1493500))
# < 왼쪽, > 오른쪽, ^ 가운데, 콤마는 천 단위 구분이야.
```

```python
# 24.5 심사문제: 특정 단어 개수 세기

text = input()
count = 0

for word in text.split():
    if word.strip(".,") == "the":
        count += 1

print(count)
```

```python
# 24.6 심사문제: 높은 가격순으로 출력하기

prices = list(map(int, input().split(";")))
prices.sort(reverse=True)

for price in prices:
    print("{:>9,}".format(price))
```
