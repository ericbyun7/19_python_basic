```python
print(len("사과"))  # 글자 2개 → 2
print(len([1, 2, 3]))  # 숫자 3개 → 3
```

```python
# len()은 몇 개 있는지 세는 함수야.

print(len("사과"))  # 글자가 2개라서 2가 나와.
print(len([1, 2, 3]))  # 숫자가 3개라서 3이 나와.
```

```python
# len()으로 문자열의 길이를 구해. 공백도 1개로 세어.
text = "안녕 친구"
print(len(text))  # 결과: 5
```

```python
# 인덱스는 글자나 항목의 위치야. 0부터 시작해!
text = "사과나무"

print(text[0])  # 사
print(text[1])  # 과
print(text[-1])  # 무: -1은 맨 마지막 위치야.
```

```python
# range()는 일정한 범위의 숫자를 차례로 꺼내게 해 줘.
# range(5)는 0부터 4까지야. 끝 숫자 5는 포함하지 않아!
for number in range(5):
    print(number)  # 0, 1, 2, 3, 4가 한 줄씩 나와.
```
