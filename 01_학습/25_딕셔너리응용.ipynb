## 25.1 setdefault: 없을 때만 추가

키가 이미 있으면 기존 값을 유지해요.

```python
# 25.1 setdefault: 없을 때만 추가

x = {'a': 10, 'b': 20, 'c': 30, 'd': 40}
x.setdefault('e')
x.setdefault('f', 100)
x.setdefault('a', 90)
print(x)
print(x['a'])  # 10
```

## 25.1 update: 수정 또는 추가

기존 키는 수정하고, 없는 키는 추가해요.

```python
# 25.1 update: 수정 또는 추가

x = {'a': 10, 'b': 20, 'c': 30, 'd': 40}
x.update(a=90, e=50)
print(x)
```

## 25.1 pop과 del: 삭제

pop은 지운 값을 돌려주고, del은 삭제만 해요.

```python
# 25.1 pop과 del: 삭제

x = {'a': 10, 'b': 20, 'c': 30, 'd': 40}
print(x.pop('a'))  # 10
del x['b']
print(x)
print(x.pop('없는키', 0))  # 없을 때 기본값 0
```

## 25.1 popitem과 clear

popitem은 마지막으로 들어온 쌍을 꺼내고, clear는 전부 지워요.

```python
# 25.1 popitem과 clear

x = {'a': 10, 'b': 20, 'c': 30, 'd': 40}
print(x.popitem())  # ('d', 40)
x.clear()
print(x)  # {}
```

## 25.1 get: 값 조회

없는 키를 대괄호로 찾으면 오류가 나요. get은 기본값을 줄 수 있어요.

```python
# 25.1 get: 값 조회

x = {'a': 10, 'b': 20, 'c': 30, 'd': 40}
print(x.get('a'))  # 10
print(x.get('e'))  # None
print(x.get('e', 0))  # 0
```

## 25.1 keys · values · items

keys는 이름표, values는 값, items는 둘의 쌍이에요.

```python
# 25.1 keys · values · items

x = {'a': 10, 'b': 20, 'c': 30, 'd': 40}
print(list(x.keys()))
print(list(x.values()))
print(list(x.items()))
```

## 25.1 fromkeys: 여러 키로 만들기

값을 생략하면 None, 값을 넣으면 같은 기본값이 들어가요.

```python
# 25.1 fromkeys: 여러 키로 만들기

keys = ['a', 'b', 'c', 'd']
print(dict.fromkeys(keys))
print(dict.fromkeys(keys, 100))
```

## 25.2 반복문으로 출력하기

딕셔너리를 그대로 반복하면 키만 나와요. items를 쓰면 값도 꺼낼 수 있어요.

```python
# 25.2 반복문으로 출력하기

x = {'a': 10, 'b': 20, 'c': 30, 'd': 40}
for key, value in x.items():
    print(key, value)

for key in x.keys():
    print(key, end=' ')
print()
for value in x.values():
    print(value, end=' ')
print()
```

## 25.3 딕셔너리 표현식

중괄호 안에서 키: 값을 반복해서 만들어요.

```python
# 25.3 딕셔너리 표현식

keys = ['a', 'b', 'c', 'd']
x = {key: value for key, value in dict.fromkeys(keys).items()}
print(x)
```

## 25.3 조건으로 골라 새 딕셔너리 만들기

반복 중 원본에서 직접 삭제하지 않고, 남길 것만 골라요.

```python
# 25.3 조건으로 골라 새 딕셔너리 만들기

x = {'a': 10, 'b': 20, 'c': 30, 'd': 40}
x = {key: value for key, value in x.items() if value != 20}
print(x)
```

## 25.4 중첩 딕셔너리

첫 대괄호로 행성을 찾고, 두 번째로 그 행성의 정보를 찾아요.

```python
# 25.4 중첩 딕셔너리

terrestrial_planet = {
    'Mercury': {'mean_radius': 2439.7, 'mass': 3.3022E+23, 'orbital_period': 87.969},
    'Venus': {'mean_radius': 6051.8, 'mass': 4.8676E+24, 'orbital_period': 224.70069},
    'Earth': {'mean_radius': 6371.0, 'mass': 5.97219E+24, 'orbital_period': 365.25641},
    'Mars': {'mean_radius': 3389.5, 'mass': 6.4185E+23, 'orbital_period': 686.9600}
}
print(terrestrial_planet['Venus']['mean_radius'])  # 6051.8
```

## 25.5 할당과 복사

y = x는 같은 딕셔너리를 가리켜요. copy는 바깥 딕셔너리를 새로 만들어요.

```python
# 25.5 할당과 복사

x = {'a': 0, 'b': 0, 'c': 0, 'd': 0}
y = x
y['a'] = 99
print(x['a'])  # 99
print(x is y)  # True

x = {'a': 0, 'b': 0, 'c': 0, 'd': 0}
y = x.copy()
y['a'] = 99
print(x['a'])  # 0
print(x is y)  # False
```

## 25.5 중첩 딕셔너리 복사

안쪽까지 독립적으로 복사하려면 deepcopy를 사용해요.

```python
# 25.5 중첩 딕셔너리 복사

import copy

x = {'a': {'python': '2.7'}, 'b': {'python': '3.6'}}
y = copy.deepcopy(x)
y['a']['python'] = '2.7.15'
print(x['a']['python'])  # 2.7
print(y['a']['python'])  # 2.7.15
```

## 25.7 연습문제: 평균 점수 구하기

값을 모두 더한 뒤 과목 수로 나눠요.

```python
# 25.7 연습문제: 평균 점수 구하기

maria = {'korean': 94, 'english': 91, 'mathematics': 89, 'science': 83}
average = sum(maria.values()) / len(maria)
print(average)  # 89.25
```

- 추가·수정: `setdefault`, `update`
- 삭제: `pop`, `del`, `clear`
- 조회: `get`, `keys`, `values`, `items`
- 반복: `for key, value in x.items()`
