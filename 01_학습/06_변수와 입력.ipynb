# 06_변수와 입력

오른쪽 위 커널은 **Python 학습 (.venv)**를 선택해. 위에서 아래로 실행해. `input()` 셀은 화면 맨 위 입력창에서 값을 기다리는 것이 정상이고, 중단하려면 ■ 버튼을 눌러.


```python
# 변수는 값을 담는 이름이야.
x = 10
# 영문, 한글, 밑줄(_)로 시작할 수 있어. 숫자로 시작할 수는 없어.
# 숫자는 첫 글자 뒤에 사용할 수 있어. 대소문자를 구분해.
# 공백, 특수문자, 파이썬 키워드는 사용할 수 없어.
```

```python
y = "hello world"
print(y)
```

```python
#변수의 자료형 알아보기

print(type(x))
print(type(y))
```

```python
# 변수 여러개를 한 번에 만들기

x, y, z = 10, 20, 30

print(x)
print(y)
print(z)
```

```python
x = y = z = 10

print(x)
print(y)
print(z)
```

```python
x, y = 10, 20
x, y = y, x
print(x)
print(y)
```

```python
# 변수로 계산하기

a = 10
b = 20
c = a + b
print(c)
```

```python
# 계산만 하면 변수 a의 값은 바뀌지 않아.
a = 10
print(a + 20)
```

```python
# 바로 위 셀을 먼저 실행해. a는 아직 10이야.
print(a)
```

```python
a = 10
a += 20 # a = a + 20
print(a)
```

```python
# 입력 값을 변수에 저장하기
input()
```

```python
x = input("문자열을 입력하세요: ")
print(x)
```

```python
a = input("첫 번째 숫자를 입력하세요: ")
b = input("두 번째 숫자를 입력하세요: ")
a = int(a)
b = int(b)
print(a + b)
```

```python
a = int(input("첫 번째 숫자를 입력하세요: "))
b = int(input("두 번째 숫자를 입력하세요: "))
print(a + b)
```

```python
# 입력 값을 변수 두 개에 저장하기

a, b = input("문자열 두 개를 입력하세요: ").split()
print(a)
print(b)
```

```python
print("hello python".split())
```

```python
a, b = ['hello', 'python']
a, b, c = 10, 20, 30
```

```python
# 두 숫자를 한 줄에 입력해. 예: 10 20
a, b = map(int, input("숫자 두 개를 입력하세요: ").split())
print(a + b)
```

```python
# map()은 각 항목에 함수를 적용해. int는 정수로 바꿔.
a, b = map(int, input("숫자 두 개를 입력하세요: ").split())
print(a + b)
```

```python
print(input("숫자 4 개를 입력하세요").split())
```

```python
a, b, c, d = map(int, input("숫자 4 개를 입력하세요: ").split())
print((a + b + c + d)/4)
```
