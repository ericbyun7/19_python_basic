```python
# 23.1 2차원 리스트를 만들고 요소에 접근하기

a = [[10, 20], [30, 40], [50, 60]]

# 위치 번호는 0부터 시작해.
#             0열  1열
# 0행         10   20
# 1행         30   40
# 2행         50   60

print(a[0])  # [10, 20]: 한 줄 전체
print(a[0][0])  # 10
print(a[1][1])  # 40
print(a[2][1])  # 60
```

```python
# 23.2.1 for 반복문을 한 번만 사용하기

a = [[10, 20], [30, 40], [50, 60]]

for x, y in a:
    print(x, y)

# 첫 번째: x=10, y=20
# 두 번째: x=30, y=40
# 세 번째: x=50, y=60
```

```python
# 23.2.2 for 반복문을 두 번 사용하기

a = [[10, 20], [30, 40], [50, 60]]

for i in a:  # 한 줄 꺼내기: [10, 20]
    for j in i:  # 그 줄의 값 꺼내기: 10, 20
        print(j, end=" ")
    print()  # 한 줄이 끝나면 줄바꿈
```

```python
# 23.2.3 for와 range 사용하기

a = [[10, 20], [30, 40], [50, 60]]

for i in range(len(a)):  # 줄 번호: 0, 1, 2
    for j in range(len(a[i])):  # 칸 번호: 0, 1
        print(a[i][j], end=" ")
    print()

# len(a) → 줄 개수 3
# len(a[i]) → 해당 줄의 칸 개수 2
```

```python
# 23.2.4 while 반복문을 한 번 사용하기

a = [[10, 20], [30, 40], [50, 60]]

i = 0

while i < len(a):
    x, y = a[i]
    print(x, y)
    i += 1
```

```python
# 23.2.5 while 반복문을 두 번 사용하기

a = [[10, 20], [30, 40], [50, 60]]

i = 0
while i < len(a):
    j = 0  # 새 줄마다 칸 번호를 0으로!
    while j < len(a[i]):
        print(a[i][j], end=" ")
        j += 1  # 다음 칸
    print()
    i += 1  # 다음 줄
```
