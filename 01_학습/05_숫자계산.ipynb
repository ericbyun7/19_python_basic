```python
# 정수계산
# 사칙연산

print(1+1)
```

```python
print(1-2)
```

```python
print(1-2)
print(1*2)
print(5/2)
```

```python
# 나눗셈 후 나머지를 구하는 %연산자

print(5%2)
```

```python
# 거듭제곱 연산자 **

print(2 ** 2)
```

```python
# 값을 정수로 만들기

# int(숫자)
# int(계산식)
# int(문자열)

print(int(3.3))
print(int(5/2))
print(int("10"))
```

```python
print(type(int(3.3)))
```

```python
# 실수계산하기

print(2.5 + 2.1)
```

```python
print(4.3 - 2.6)
print(4.3 * 2.6)
print(4.3 / 2.6)
```

```python
# 실수와 정수 함께 계산

print(4.2 + 5)
```

```python
# 값을 실수로 만들기

print(type(float(5)))
print(float(1+2))
print(float(5))
```

```python
# 곱셈을 먼저 계산해. 괄호가 있으면 괄호부터 계산해.
print(35 + 1 * 2)  # 37
print((35 + 1) * 2)  # 72
```
