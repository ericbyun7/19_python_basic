**인수**는 함수를 호출할 때 넣는 값, **매개변수**는 그 값을 받는 이름이에요. 각 코드 셀을 따로 실행해도 됩니다.

## 30.1 위치 인수

값을 넣은 순서대로 a, b, c가 받아요.

```python
# 30.1 위치 인수

def print_numbers(a, b, c):
    print(a)
    print(b)
    print(c)

print_numbers(10, 20, 30)
```

## 30.1 리스트·튜플 언패킹

호출할 때 *는 묶인 값을 풀어 넣어요. 받는 칸은 세 개이므로 값도 세 개 필요해요.

```python
# 30.1 리스트·튜플 언패킹

def print_numbers(a, b, c):
    print(a, b, c)

x = [10, 20, 30]
print_numbers(*x)
print_numbers(*(10, 20, 30))
```

## 30.1 가변 위치 인수: *args

정의할 때 *args는 여러 값을 튜플로 모아 받아요.

```python
# 30.1 가변 위치 인수: *args

def print_numbers(*args):
    for arg in args:
        print(arg)

print_numbers(10)
print_numbers(10, 20, 30, 40)
```

## 30.1 고정 인수와 가변 인수

첫 값은 a가 받고, 나머지는 args에 모여요.

```python
# 30.1 고정 인수와 가변 인수

def print_numbers(a, *args):
    print(a)
    print(args)

print_numbers(1, 10, 20)  # a=1, args=(10, 20)
```

## 30.2 키워드 인수

이름=값으로 넣으면 순서를 바꿔도 돼요.

```python
# 30.2 키워드 인수

def personal_info(name, age, address):
    print("이름:", name)
    print("나이:", age)
    print("주소:", address)

personal_info(age=30, address="서울시 용산구 이촌동", name="홍길동")
```

## 30.3 딕셔너리 언패킹

호출할 때 **는 키를 매개변수 이름에 맞춰 값을 전달해요.

```python
# 30.3 딕셔너리 언패킹

def personal_info(name, age, address):
    print(name, age, address)

x = {"name": "홍길동", "age": 30, "address": "서울시 용산구 이촌동"}
personal_info(**x)
```

## 30.3 가변 키워드 인수: **kwargs

정의할 때 **kwargs는 이름 붙은 값들을 딕셔너리로 모아요.

```python
# 30.3 가변 키워드 인수: **kwargs

def personal_info(**kwargs):
    for key, value in kwargs.items():
        print(key, value)

personal_info(name="홍길동", age=30, address="서울시 용산구 이촌동")
```

## 30.3 *args와 **kwargs 함께 받기

이름 없이 넣은 값과 이름 붙인 값을 나눠 받아요.

```python
# 30.3 *args와 **kwargs 함께 받기

def custom_print(*args, **kwargs):
    print(*args, **kwargs)

custom_print(1, 2, 3, sep=":", end="\n")
```

## 30.4 매개변수 초깃값

초깃값이 있으면 그 인수는 생략해도 돼요. 초깃값 없는 매개변수를 먼저 적어요.

```python
# 30.4 매개변수 초깃값

def personal_info(name, age, address="비공개"):
    print(name, age, address)

personal_info("홍길동", 30)
personal_info("홍길동", 30, "서울시 용산구 이촌동")
```

## 30.6 연습문제: 가장 높은 점수

여러 점수 중 가장 큰 값을 찾아 반환해요.

```python
# 30.6 연습문제: 가장 높은 점수

def get_max_score(*args):
    return max(args)

korean, english, mathematics, science = 100, 86, 81, 91
print(get_max_score(korean, english, mathematics, science))  # 100
print(get_max_score(english, science))  # 91
```

## 한눈에 보기

| 사용 위치 | 뜻 |
|---|---|
| 호출 `f(*x)` | 리스트·튜플 풀기 |
| 정의 `def f(*args)` | 위치 인수를 튜플로 모으기 |
| 호출 `f(**x)` | 딕셔너리를 이름=값으로 풀기 |
| 정의 `def f(**kwargs)` | 키워드 인수를 딕셔너리로 모으기 |
