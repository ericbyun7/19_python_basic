```python
# 20번 반복: i는 0부터 19까지 바뀌어.

i = 0  # 시작 숫자

while i < 20:  # i가 20보다 작은 동안 반복
    print("Hello, world!", i)
    i += 1  # i를 1 증가시켜. i = i + 1

# 결과: Hello, world! 0부터 Hello, world! 19까지 출력
```

```python
i = 30

while i > 0:
    print(i)
    i -= 1  # i = i - 1

# 결과: 30부터 1까지 한 줄씩 출력
```

```python
# 실행 후 맨 위 입력칸에 3을 입력.

count = int(input())

i = 0

while i < count:
    print("안녕!")
    i += 1

# 3을 입력하면 안녕!이 3번 출력.
```
