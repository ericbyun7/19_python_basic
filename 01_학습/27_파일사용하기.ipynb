# 유닛 27. 파일 사용하기

파일은 프로그램이 끝나도 내용을 남겨 두는 저장 공간이에요. 각 코드 셀을 따로 실행할 수 있어요. 셀마다 실습 폴더를 준비하고, 읽을 파일이 없으면 예제 파일도 만들어요. 기존 파일은 읽기 준비 과정에서 덮어쓰지 않아요.

교재 예제를 바탕으로 한글 저장을 위한 `encoding="utf-8"`을 추가했어요. 실습 파일은 현재 작업 폴더 아래 `27_실습` 폴더에 모아요.

`w`는 기존 내용을 새로 쓰고, `r`은 읽어요. `with`를 사용하면 파일을 자동으로 닫아요.

## 실습 준비

이 셀부터 실행하세요. 실제 저장 위치가 출력돼요.

```python
# 실습 준비

from pathlib import Path

practice = Path("27_실습")
practice.mkdir(exist_ok=True)
print(practice.resolve())
```

## 27.1.1 파일에 문자열 쓰기

open으로 열고, write로 쓰고, close로 닫아요.

```python
# 27.1.1 파일에 문자열 쓰기

from pathlib import Path

practice = Path("27_실습")
practice.mkdir(exist_ok=True)

file = open(practice / "hello.txt", "w", encoding="utf-8")
file.write("Hello, world!")
file.close()
```

## 27.1.2 파일에서 문자열 읽기

read는 파일 내용을 하나의 문자열로 가져와요.

```python
# 27.1.2 파일에서 문자열 읽기

from pathlib import Path

practice = Path("27_실습")
practice.mkdir(exist_ok=True)

# 파일이 없을 때만 예제 내용을 준비해.
if not (practice / "hello.txt").exists():
    (practice / "hello.txt").write_text("Hello, world!", encoding="utf-8")

file = open(practice / "hello.txt", "r", encoding="utf-8")
s = file.read()
print(s)  # Hello, world!
file.close()
```

## 27.1.3 with로 자동으로 닫기

들여쓰기 블록이 끝나면 파일도 닫혀요.

```python
# 27.1.3 with로 자동으로 닫기

from pathlib import Path

practice = Path("27_실습")
practice.mkdir(exist_ok=True)

# 파일이 없을 때만 예제 내용을 준비해.
if not (practice / "hello.txt").exists():
    (practice / "hello.txt").write_text("Hello, world!", encoding="utf-8")

with open(practice / "hello.txt", "r", encoding="utf-8") as file:
    print(file.read())
```

## 27.2.1 반복문으로 여러 줄 쓰기

문자열 끝의 \n이 줄바꿈이에요.

```python
# 27.2.1 반복문으로 여러 줄 쓰기

from pathlib import Path

practice = Path("27_실습")
practice.mkdir(exist_ok=True)

with open(practice / "hello.txt", "w", encoding="utf-8") as file:
    for i in range(3):
        file.write("Hello, world! {0}\n".format(i))
```

## 27.2.2 리스트의 문자열 쓰기

writelines는 줄바꿈을 자동으로 넣어 주지 않아요.

```python
# 27.2.2 리스트의 문자열 쓰기

from pathlib import Path

practice = Path("27_실습")
practice.mkdir(exist_ok=True)

lines = ["안녕하세요.\n", "파이썬\n", "코딩 도장입니다.\n"]
with open(practice / "hello.txt", "w", encoding="utf-8") as file:
    file.writelines(lines)
```

## 27.2.3 여러 줄을 리스트로 읽기

readlines는 줄별 문자열을 리스트로 가져와요.

```python
# 27.2.3 여러 줄을 리스트로 읽기

from pathlib import Path

practice = Path("27_실습")
practice.mkdir(exist_ok=True)

# 파일이 없을 때만 읽기 예제를 준비해.
if not (practice / "hello.txt").exists():
    (practice / "hello.txt").write_text("안녕하세요.\n파이썬\n코딩 도장입니다.\n", encoding="utf-8")

with open(practice / "hello.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()
    print(lines)
```

## 27.2.4 readline과 while

readline은 한 줄씩 읽고, 더 읽을 내용이 없으면 빈 문자열을 돌려줘요.

```python
# 27.2.4 readline과 while

from pathlib import Path

practice = Path("27_실습")
practice.mkdir(exist_ok=True)

# 파일이 없을 때만 읽기 예제를 준비해.
if not (practice / "hello.txt").exists():
    (practice / "hello.txt").write_text("안녕하세요.\n파이썬\n코딩 도장입니다.\n", encoding="utf-8")

with open(practice / "hello.txt", "r", encoding="utf-8") as file:
    line = None
    while line != "":
        line = file.readline()
        print(line.strip("\n"))
# 교재 방식은 마지막에 빈 줄도 한 번 출력해요.
```

## 27.2.5 for로 한 줄씩 읽기

파일을 for에 넣으면 줄을 하나씩 꺼내요.

```python
# 27.2.5 for로 한 줄씩 읽기

from pathlib import Path

practice = Path("27_실습")
practice.mkdir(exist_ok=True)

# 파일이 없을 때만 읽기 예제를 준비해.
if not (practice / "hello.txt").exists():
    (practice / "hello.txt").write_text("안녕하세요.\n파이썬\n코딩 도장입니다.\n", encoding="utf-8")

with open(practice / "hello.txt", "r", encoding="utf-8") as file:
    for line in file:
        print(line.strip("\n"))
```

## 27.3.1 객체 저장하기: pickle.dump

리스트나 딕셔너리도 파일에 저장할 수 있어요. wb는 바이너리 쓰기예요.

```python
# 27.3.1 객체 저장하기: pickle.dump

from pathlib import Path

practice = Path("27_실습")
practice.mkdir(exist_ok=True)

import pickle

name = "james"
age = 17
address = "서울시 서초구 반포동"
scores = {"korean": 90, "english": 95, "mathematics": 85, "science": 82}

with open(practice / "james.p", "wb") as file:
    pickle.dump(name, file)
    pickle.dump(age, file)
    pickle.dump(address, file)
    pickle.dump(scores, file)
```

## 27.3.2 객체 읽기: pickle.load

저장한 순서대로 읽어요. 여기서 직접 만든 james.p 파일을 사용하세요.

```python
# 27.3.2 객체 읽기: pickle.load

from pathlib import Path

practice = Path("27_실습")
practice.mkdir(exist_ok=True)

import pickle

# 파일이 없을 때만 예제 데이터를 저장해.
if not (practice / "james.p").exists():
    with open(practice / "james.p", "wb") as sample:
        pickle.dump("james", sample)
        pickle.dump(17, sample)
        pickle.dump("서울시 서초구 반포동", sample)
        pickle.dump({"korean": 90, "english": 95, "mathematics": 85, "science": 82}, sample)

with open(practice / "james.p", "rb") as file:
    name = pickle.load(file)
    age = pickle.load(file)
    address = pickle.load(file)
    scores = pickle.load(file)

print(name)
print(age)
print(address)
print(scores)
```

## 27.5 연습 준비: words.txt 만들기

교재에서 제공하는 단어 파일을 실습 폴더에 만들어요.

```python
# 27.5 연습 준비: words.txt 만들기

from pathlib import Path

practice = Path("27_실습")
practice.mkdir(exist_ok=True)

words = [
    "anonymously",
    "compatibility",
    "dashboard",
    "experience",
    "photography",
    "spotlight",
    "warehouse",
]

with open(practice / "words.txt", "w", encoding="utf-8") as file:
    for word in words:
        file.write(word + "\n")
```

## 27.5 연습문제: 10자 이하인 단어 개수

줄바꿈 문자를 빼고 길이를 세어요.

```python
# 27.5 연습문제: 10자 이하인 단어 개수

from pathlib import Path

practice = Path("27_실습")
practice.mkdir(exist_ok=True)

# 파일이 없을 때만 교재의 단어 목록을 준비해.
if not (practice / "words.txt").exists():
    (practice / "words.txt").write_text("anonymously\ncompatibility\ndashboard\nexperience\nphotography\nspotlight\nwarehouse\n", encoding="utf-8")

with open(practice / "words.txt", "r", encoding="utf-8") as file:
    count = 0
    words = file.readlines()
    for word in words:
        if len(word.strip("\n")) <= 10:
            count += 1
    print(count)  # 4
```

## 기억하기

| 기능 | 사용법 |
|---|---|
| 쓰기 | `write()` |
| 전체 읽기 | `read()` |
| 리스트로 쓰기 | `writelines()` |
| 줄별 리스트로 읽기 | `readlines()` |
| 한 줄 읽기 | `readline()` |
| 자동으로 닫기 | `with open(...) as file:` |
| 객체 저장·읽기 | `pickle.dump()` / `pickle.load()` |

읽기 셀은 파일이 없을 때만 예제 데이터를 준비해요. 이미 파일이 있으면 그 내용을 읽어요. 27.4 퀴즈·27.6 심사문제는 별도입니다.
