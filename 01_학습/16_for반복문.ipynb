```python
# 교재는 100번 반복해.
# 먼저 3번만 실행해 보자.

for i in range(3):
    print("Hello, world!")

# 결과: Hello, world!가 3줄 출력돼.
# 반복할 코드는 공백 4칸 들여쓰기!
```

```python
# i는 숫자를 하나씩 받는 변수야.
# range(3)에서는 0, 1, 2를 차례로 받아.

for i in range(3):
    print(i)

# 결과:
# 0
# 1
# 2
```

```python
# range(시작, 끝): 끝 숫자는 포함하지 않아.

for i in range(5, 8):
    print(i)

# 결과: 5, 6, 7이 한 줄씩 출력돼.
```

```python
# range(시작, 끝, 증가폭)
# 0부터 시작해서 2씩 커져.

for i in range(0, 10, 2):
    print(i)

# 결과: 0, 2, 4, 6, 8
```

```python
# -1은 숫자를 1씩 줄이라는 뜻이야.

for i in range(5, 0, -1):
    print(i)

# 결과: 5, 4, 3, 2, 1
# 끝 숫자인 0은 포함하지 않아.
```

```python
# 실행 후 맨 위 입력칸에 3을 입력하고 Enter!

count = int(input())

for i in range(count):
    print("안녕!")

# 3을 입력하면 안녕!이 3번 출력돼.
```

```python
# 리스트에서 값을 하나씩 꺼내.

numbers = [10, 20, 30]

for number in numbers:
    print(number)

# 결과: 10, 20, 30
# 변수 이름은 i가 아니어도 돼.
```

```python
# 튜플도 하나씩 꺼낼 수 있어.

fruits = ("apple", "orange", "grape")

for fruit in fruits:
    print(fruit)
```

```python
# 문자열에서는 글자를 하나씩 꺼내.
# end=" "는 줄바꿈 대신 공백을 넣어.

for letter in "Python":
    print(letter, end=" ")

# 결과: P y t h o n
```

```python
numbers = [10, 20, 30]

for number in numbers:
    print(number)
```

```python
for letter in "나비":
    print(letter)

# 나
# 비
```

```python
# 16.5 연습문제

x = [49, -17, 25, 102, 8, 62, 21]

for i in x:
    print(i * 10, end=" ")

# end=" "는 줄바꿈 대신 공백을 넣어.
```

```python
# 16.6 심사문제
# 실행 후 맨 위 입력칸에 단 숫자를 입력해.

n = int(input())

for i in range(1, 10):
    print(n, "*", i, "=", n * i)
```
