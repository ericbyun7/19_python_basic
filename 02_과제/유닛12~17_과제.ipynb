```python
# 13
price = int(input())
coupon = input()

if coupon == "Cash3000":
    price -= 3000
if coupon == "Cash5000":
    price -= 5000

print(price)
```

```python
# 14
korean, english, mathematics, science = map(int, input().split())

if (
    0 <= korean <= 100
    and 0 <= english <= 100
    and 0 <= mathematics <= 100
    and 0 <= science <= 100
):
    average = (korean + english + mathematics + science) / 4

    if average >= 80:
        print("합격")
    else:
        print("불합격")
else:
    print("잘못된 점수")
```

```python
# 15
age = int(input())
balance = 9000

if age <= 12:
    balance -= 650
elif age <= 18:
    balance -= 1050
else:
    balance -= 1250

print(balance)
```
