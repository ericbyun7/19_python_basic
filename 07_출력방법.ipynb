# 07_출력방법

오른쪽 위 커널은 **Python 학습 (.venv)**를 선택해. 위에서 아래로 실행해. `input()` 셀은 화면 맨 위 입력창에서 값을 기다리는 것이 정상이고, 중단하려면 ■ 버튼을 눌러.


```python
# 값을 여러개 출력하기
print(1, 2, 3)
```

```python
print('hello', 'world', 'python')
```

```python
print(1, 2, 3, sep=',') # sep 옵션으로 구분자 지정
```

```python
print(4, 5, 6, sep='-', end='')
```

```python
print(1, end='')
print(2, end='')
print(3)
```

```python
# 연도 월 일 시 분 초를 한 줄에 입력해. 예: 2000 10 27 11 43 59
year, month, day, hour, minute, second = input().split()
```

```python
print(year, month, day, sep='-')
print(hour, minute, second, sep=':')
```
